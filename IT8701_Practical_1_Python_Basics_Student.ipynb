## Question 2 — Apply delivery and check the budget
Continue with the one-meal order from Question 1.

1. Use `if/else` to set `delivery_cost` according to the delivery rule.
2. Add delivery to get `final_cost`.
3. Use `if/else` to print the amount remaining when the final cost is within budget, or the amount over budget otherwise. An order costing exactly the budget is within budget.
4. Print the subtotal, delivery cost, and final cost before the budget message.

**Expected output with quantity `2`:**
```text
Subtotal: $17.00
Delivery: $5.00
Final cost: $22.00
Within budget: $18.00 remaining
```
**Try the boundary:** temporarily replace `subtotal = line_cost` with `subtotal = 39`, then with `subtotal = 40`, and run your answer each time. Check delivery is $5 and $0 respectively. Restore `subtotal = line_cost` before the automatic check.


````python
budget = 40.00
delivery_fee = 5.00
free_delivery_threshold = 40.00
subtotal = line_cost

# Delivery is charged only when the subtotal is below $40
if subtotal < free_delivery_threshold:
    delivery_cost = delivery_fee
else:
    delivery_cost = 0.00

final_cost = subtotal + delivery_cost

print(f"Subtotal: ${subtotal:.2f}")
print(f"Delivery: ${delivery_cost:.2f}")
print(f"Final cost: ${final_cost:.2f}")

# An order costing exactly the budget counts as within budget, so use <=
if final_cost >= budget:
    print(f"Within budget: ${budget - final_cost:.2f} remaining")
else:
    print(f"Over budget: ${final_cost - budget:.2f} over")
````

## Question 3 — Summarise several meals
The team now orders three types of meal. Use the supplied list below; it replaces the single meal from Questions 1–2.

Each dictionary holds one meal's name, unit price, and quantity. For example, `order["quantity"]` accesses the quantity in the current dictionary.

1. Set `total_quantity` to `0` and `subtotal` to `0.0` **before** the loop.
2. Complete the loop to calculate and print each meal's line cost.
3. Add each quantity to `total_quantity` and each line cost to `subtotal`.
4. After the loop, print total quantity and subtotal.

**Hint:** `running_total += value` adds a value to a running total. Indented lines run once per meal; lines after the loop run once.

**Expected output:**
```text
Rice bowl x 2: $17.00
Noodles x 1: $7.00
Salad x 1: $9.00
Total quantity: 4
Subtotal: $33.00
```


````python
orders = [
    {"meal": "Rice bowl", "unit_price": 8.50, "quantity": 2},
    {"meal": "Noodles", "unit_price": 7.00, "quantity": 1},
    {"meal": "Salad", "unit_price": 9.00, "quantity": 1},
]

# Start both totals at zero, once, before the loop
total_quantity = 0
subtotal = 0.0

for order in orders:
    line_cost = order["unit_price"] * order["quantity"]
    print(f"{order['meal']} x {order['quantity']}: ${line_cost:.2f}")
    total_quantity += order["quantity"]
    subtotal += line_cost

# After the loop: print the final totals once
print(f"Total quantity: {total_quantity}")
print(f"Subtotal: ${subtotal:.2f}")
````

**Check your work:** run the completed receipt cell first.


````python
assert "total_quantity" in globals(), "Create total_quantity before the loop."
assert total_quantity == 4, "Add every meal's quantity; initialise the total before the loop."
assert abs(subtotal - 33.00) < 0.001, "Add all line costs; do not reset subtotal inside the loop."
print("Q3 checks passed. Compare all five printed lines with the example.")
````

## Question 4 — Make the calculation reusable
Move the delivery calculation into a function so it can be used for different orders.

1. Complete `calculate_total(subtotal, delivery_fee, free_delivery_threshold)`. Use its parameters to apply delivery and **return** the final cost. Replace the placeholder `return None`.
2. Call it with Question 3's subtotal, a $5 fee, and a $40 threshold. Store the result as `final_cost`.
3. Print the final cost and the budget result, using a $40 budget.

**Hint:** `return` sends a value back to the caller; `print()` displays it. This function should return a number. Put the printing outside the function.

**Expected output for the three-meal order:**
```text
Final cost: $38.00
Within budget: $2.00 remaining
```


````python
def calculate_total(subtotal, delivery_fee, free_delivery_threshold):
    """Return the subtotal plus any delivery charge."""
    if subtotal < free_delivery_threshold:
        return subtotal + delivery_fee
    else:
        return subtotal

# Q3 subtotal, $5 fee, $40 threshold
final_cost = calculate_total(subtotal, delivery_fee, free_delivery_threshold)

print(f"Final cost: ${final_cost:.2f}")
if final_cost <= budget:
    print(f"Within budget: ${budget - final_cost:.2f} remaining")
else:
    print(f"Over budget: ${final_cost - budget:.2f} over")
````

**Check your work:** these examples check below, at, and above the threshold. The final example checks that your function uses its parameters.


````python
assert calculate_total(33, 5, 40) == 38, "Add delivery below the threshold; return the number."
assert calculate_total(40, 5, 40) == 40, "Delivery is free at the threshold."
assert calculate_total(45, 5, 40) == 45, "Delivery is free above the threshold."
assert calculate_total(20, 3, 25) == 23, "Use the supplied fee and threshold parameters."
assert abs(subtotal - 33.00) < 0.001, "Rerun Question 3 before this question."
assert final_cost == 38, "Store the returned result for the three-meal order in final_cost."
print("Q4 checks passed. Check the printed budget message too.")
````
